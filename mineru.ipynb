```python
from google.colab import files
uploaded = files.upload()
pdf_path = list(uploaded.keys())[0]
print("Uploaded:", pdf_path)
```

```python
import pymupdf
from PIL import Image

doc = pymupdf.open(pdf_path)
pages = []
for page in doc:
    pix = page.get_pixmap(dpi=200)
    img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
    pages.append(img)
print(len(pages), "pages ready")
```

```python
from transformers import AutoProcessor, Qwen2VLForConditionalGeneration
from mineru_vl_utils import MinerUClient

name = "opendatalab/MinerU2.5-Pro-2604-1.2B"
model = Qwen2VLForConditionalGeneration.from_pretrained(
    name, dtype="auto", device_map="auto"
)
processor = AutoProcessor.from_pretrained(name, use_fast=True)

client = MinerUClient(
    backend="transformers", model=model, processor=processor,
    image_analysis=False
)
```

```python
from mineru_vl_utils.post_process import json2md

blocks = client.two_step_extract(pages[27])
print(json2md(blocks))
```
